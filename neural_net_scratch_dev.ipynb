```python
"""
NETWORK WITH A BIAS - now it can learn 2x + 1.

Two changes from before:
  1. forward pass adds the bias:        prediction = x * weight + bias
  2. we update the bias too, with slope: 2 * error  (no input factor,
     because the bias isn't tied to any input -- it just adds itself).
"""
import random

# Rule: output = 2*input + 1  (still hidden from the network)
data = [(1.0, 3.0), (2.0, 5.0), (3.0, 7.0), (4.0, 9.0)]

# Two things to learn now: a weight AND a bias. Both start random.
weight = random.uniform(-1.0, 1.0)
bias   = random.uniform(-1.0, 1.0)
print(f"Start -> weight: {weight:.4f}, bias: {bias:.4f}")
print("Target -> weight should become 2.0, bias should become 1.0\n")

learning_rate = 0.01
epochs = 1000

for epoch in range(epochs):
    total_error_squared = 0.0
    for x, correct in data:

        # --- FORWARD PASS (now with bias) ---
        prediction = x * weight + bias

        # --- ERROR ---
        error = prediction - correct
        total_error_squared += error ** 2

        # --- SLOPES (gradients) ---
        weight_slope = 2 * error * x   # weight is tied to the input
        bias_slope   = 2 * error       # bias is tied to nothing -> no input factor

        # --- STEP BOTH DOWNHILL ---
        weight = weight - learning_rate * weight_slope
        bias   = bias   - learning_rate * bias_slope

    if epoch % 100 == 0:
        print(f"Epoch {epoch:4d} | weight = {weight:.4f} | bias = {bias:.4f} | error = {total_error_squared:.4f}")

print(f"\nFinal -> weight: {weight:.4f}, bias: {bias:.4f}")
print("Predictions vs correct answers:")
for x, correct in data:
    print(f"  input {x}: predicted {x*weight + bias:.2f}, correct {correct}")
print(f"\nTest unseen input 10 -> {10*weight + bias:.2f}  (should be ~21)")
```

A sigmoid function that takes real-value input and turns it into an S-shaped graph with the out of that graph being between 0-1. Neural networks need it to introduce nonlinearity. Thus allowing it to model complex data and convert the raw output into probabilities for binary classification. Unlike linear activation functions, sigmoid allows neural networks to learn non-linear patterns in data, which is necessary for complex, real-world problems.
