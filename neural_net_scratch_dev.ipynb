```python
"""
NETWORK WITH A BIAS - now it can learn 2·x₁ + 3·x₂ + 1·x₃ + 5 .
Two changes from before:
  1. forward pass adds the bias:        prediction = x * weight + bias
  2. we update the bias too, with slope: 2 * error  (no input factor,
     because the bias isn't tied to any input -- it just adds itself).
"""
import random
# Rule: output = 2·x₁ + 3·x₂ + 1·x₃ + 5 (still hidden from the network)
data = [(1.0,1.0,1.0,11.0),(2.0,0.0,1.0,10.0),(0.0,2.0,1.0,12.0),(1.0,1.0,0.0,10.0)]

# Two things to learn now: a weight AND a bias. Both start random.
weights = [random.uniform(-1.0, 1.0) for _ in range(3)] #making one weight variable like an array in js , where one variable can hold more than one thing (in this case it will be the 3 weights)
bias   = random.uniform(-1.0, 1.0)



learning_rate = 0.01
epochs = 2000

for epoch in range(epochs):
    total_error_squared = 0.0
    for x1 ,x2 , x3 , correct in data:

      # --- FORWARD PASS (now with bias) ---
          prediction = x1 * weights[0] + x2 * weights[1] + x3 * weights[2] + bias # the square bralets show the weight at each position, position 0 is weight 1

      # --- ERROR ---
          error = prediction - correct
          total_error_squared += error ** 2

      # --- SLOPES (gradients) ---
          w0_slope = 2 * error * x1   # weight is tied to the input number 1
          w1_slope = 2 * error * x2   # weight is tied to the input number 2
          w2_slope = 2 * error * x3   # weight is tied to the input number 3
          bias_slope   = 2 * error       # bias is tied to nothing -> no input factor

      # --- STEP BOTH DOWNHILL ---
          weights[0] = weights[0] - learning_rate * w0_slope
          weights[1] = weights[1] - learning_rate * w1_slope
          weights[2] = weights[2] - learning_rate * w2_slope
          bias   = bias   - learning_rate * bias_slope

if epoch % 200 == 0:
        rounded = [round(w, 3) for w in weights]
        print(f"Epoch {epoch:4d} | weights = {rounded} | bias = {bias:.3f} | error = {total_error_squared:.4f}")

        print(f"\nFinal weights = {[round(w,3) for w in weights]}, bias = {bias:.3f}")
        print("Target       = [2.0, 3.0, 1.0], bias = 5.0\n")
for x1, x2, x3, correct in data:
    pred = x1*weights[0] + x2*weights[1] + x3*weights[2] + bias
    print(f"  inputs ({x1},{x2},{x3}): predicted {pred:.2f}, correct {correct}")
```
