```python
"""
YOUR FIRST NEURAL NETWORK - from scratch, no libraries.

Goal: teach a network to learn that output = 2 * input.
We never tell it the answer is 2. It must discover the weight on its own.

This uses ONE input and ONE weight. That's the simplest learning machine
that can exist. Master this and everything else is just more of the same.
"""

import random

# ----------------------------------------------------------------------
# STEP 1: The training data.
# Each pair is (input, correct_answer). The correct answer is always 2x.
# The network can only SEE these pairs. It cannot see the rule "times 2".
# ----------------------------------------------------------------------
data = [
    (1.0, 2.0),
    (2.0, 4.0),
    (3.0, 6.0),
    (4.0, 8.0),
]

# ----------------------------------------------------------------------
# STEP 2: Start the weight at a RANDOM guess.
# Remember: we never start at 0. We start random and let learning fix it.
# ----------------------------------------------------------------------
weight = random.uniform(-1.0, 1.0)
print(f"Starting weight (random guess): {weight:.4f}")
print(f"The network must discover that this should become 2.0\n")

# ----------------------------------------------------------------------
# STEP 3: The learning rate.
# This controls how BIG a step we take downhill each time.
# Too big = we overshoot and bounce around. Too small = learning crawls.
# 0.01 is a safe small step.
# ----------------------------------------------------------------------
learning_rate = 0.01

# ----------------------------------------------------------------------
# STEP 4: The training loop.
# We repeat the whole process many times. Each full pass over the data
# is called an "epoch".
# ----------------------------------------------------------------------
epochs = 100

for epoch in range(epochs):
    total_error_squared = 0.0  # track how wrong we are this epoch

    for x, correct in data:

        # --- FORWARD PASS: make a prediction ---
        # This is the neuron you already understand: input * weight.
        prediction = x * weight

        # --- MEASURE THE ERROR ---
        error = prediction - correct        # how far off, with sign
        total_error_squared += error ** 2   # squared, for reporting

        # --- THE SLOPE (gradient) ---
        # The loss is error^2. Slope of error^2 is (2 * error).
        # But error itself depends on the weight through (x * weight),
        # so the slope with respect to the WEIGHT is: 2 * error * x.
        # (The x comes from the input being multiplied by the weight.)
        slope = 2 * error * x

        # --- STEP DOWNHILL ---
        # Move the weight OPPOSITE to the slope (that's the minus sign).
        # Scaled by the learning rate so we take a small, safe step.
        weight = weight - learning_rate * slope

    # print progress every 10 epochs so we can watch it learn
    if epoch % 10 == 0:
        print(f"Epoch {epoch:3d} | weight = {weight:.4f} | total error = {total_error_squared:.4f}")

# ----------------------------------------------------------------------
# STEP 5: The result.
# ----------------------------------------------------------------------
print(f"\nFinal weight: {weight:.4f}  (target was 2.0)")
print(f"Test: input 10 -> prediction {10 * weight:.4f}  (should be ~20)")
```

A sigmoid function that takes real-value input and turns it into an S-shaped graph with the out of that graph being between 0-1. Neural networks need it to introduce nonlinearity. Thus allowing it to model complex data and convert the raw output into probabilities for binary classification. Unlike linear activation functions, sigmoid allows neural networks to learn non-linear patterns in data, which is necessary for complex, real-world problems.
